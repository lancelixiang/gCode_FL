```python
from phe import paillier

# 生成公钥和私钥
public_key, private_key = paillier.generate_paillier_keypair()

print("公钥:", public_key)
print("私钥:", private_key)
```

```python
# 要加密的数字
secret_number = 42

# 使用公钥加密
encrypted_number = public_key.encrypt(secret_number)
print("加密后的值:", encrypted_number)

# 使用私钥解密
decrypted_number = private_key.decrypt(encrypted_number)
print("解密后的值:", decrypted_number)
```

```python
# 加密两个数字
a = 2.8403e-06
b = -2.2962e-06
encrypted_1 = public_key.encrypt(a)
encrypted_2 = public_key.encrypt(b)

# 加密数字相加
encrypted_sum = encrypted_1*2 + encrypted_2/2

# 解密结果
decrypted_sum = private_key.decrypt(encrypted_sum)
print(a*2 + b/2)
print(a, " + ", b, " =", decrypted_sum)  # 输出30
```
